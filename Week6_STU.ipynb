### 1.3 IQR check — one column first
The interquartile range (IQR) is `Q3 - Q1`. Values below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR` are flagged as potential outliers.

Start with `Systolic_BP_mmHg` (this is the **actual name** in your file).

```python
col = 'Systolic_BP_mmHg'
q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

outliers = df[(df[col] < lower_fence) | (df[col] > upper_fence)]
print(f"Lower fence: {lower_fence:.2f}")
print(f"Upper fence: {upper_fence:.2f}")
print(f"Outlier count in {col}: {len(outliers)}")

```

```python
results = []

for col in df.select_dtypes(include='number').columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    n_out = ((df[col] < q1 - 1.5 * iqr) |
             (df[col] > q3 + 1.5 * iqr)).sum()

    # TODO: write the 1% condition
    percentage = n_out / len(df) * 100
    if percentage > 1:
        status = 'WARNING'
    else:
        status = 'OK'

    results.append([col, n_out, percentage, status])

pd.DataFrame(results, columns=['Column', 'Outlier Count', 'percentage','Status'])
```

## Task 2 — Clean the data with a reusable function (20 min)
**Aim:** Use a dictionary, `for` loop and `if / elif / else` to fill gaps in a **copy** of the data.

### 2.1 Choose cleaning strategies
Numerical columns → **median**; `Salt_Intake` and `Stress_Level` → **mode**; missing `Alcohol_Consumption` → **'Unknown'** (not `'None'`, which could falsely imply non-drinking).

The dictionary is provided so you can focus on writing the reusable function. **Check:** 8 columns have gaps.


```python
strategies = {
    'Waist_Circumference_cm': 'median',
    'Sleep_Hours': 'median',
    'Creatinine_mg_dL': 'median',
    'Fasting_Glucose_mg_dL': 'median',
    'Cholesterol_mg_dL': 'median',
    'Salt_Intake': 'mode',
    'Stress_Level': 'mode',
    'Alcohol_Consumption': 'Unknown'
}

print('Columns to clean:', len(strategies))
```

### 2.2 Complete `clean_data(data)`
Use the `strategies` dictionary to fill gaps on a copy of the data, then return the cleaned DataFrame.

```python

def clean_data(data):
    data = data.copy()

    for column, method in strategies.items():
        if method == 'median':
            data[column] = data[column].fillna(data[column].median())

        elif method == 'mode':
            # fill using mode
            data[column] = data[column].fillna(data[column].mode()[0])

        else:
            # fill using method
            data[column] = data[column].fillna(method=method)

    return data
```

### 2.3 Test your function

```python
df_clean = clean_data(df)
print('Original missing values:', df.isnull().sum().sum())
print('Cleaned missing values:', df_clean.isnull().sum().sum())
print('Cleaned shape:', df_clean.shape)
```
