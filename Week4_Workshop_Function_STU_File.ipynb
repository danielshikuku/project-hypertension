## Activity 1 — A Function That Profiles a Numeric Column

**Task:** Write a function `profile_numeric(df, column_name)` that calculates four things for any given column: how many non-missing values it has, its mean, its minimum, and its maximum.

The function takes two inputs:

`df` — the dataset

`column_name` — the name of the column you want to check (e.g. 'BMI', 'Age')

Test your function on the 'BMI' column.

Output:

BMI -> count: 8045, mean: 27.03, min: 15.0, max: 55.0

````python
# Solution:
def profile_numeric(df, column_name):
    column = df[column_name]
    count = column.count()          # non-missing values only
    mean_val = column.mean()
    min_val = column.min()
    max_val = column.max()
    return count, mean_val, min_val, max_val

# Test:
count, mean_val, min_val, max_val = profile_numeric(df, 'BMI')
print(f"BMI -> count: {count}, mean: {mean_val:.2f}, min: {min_val}, max: {max_val}")
````

## Activity 2 — A Function That Profiles a Categorical Column

**Task:** Task: Write a function `profile_categorical(df, column_name)` that calculates three things for any given categorical column:
* how many values are missing,
* how many unique categories exist, and
* which category is the most common.

Test your function on the 'Smoking_Status' column.

Output:

Smoking_Status -> missing: 0, unique categories: 3, most common: Never

````python
# Solution:
def profile_categorical(df, column_name):
    column = df[column_name]
    missing = column.isnull().sum()
    unique_count = column.nunique()
    most_common = column.mode()[0]
    return missing, unique_count, most_common

# Test:
missing, unique_count, most_common = profile_categorical(df, 'Smoking_Status')
print(f"Smoking_Status -> missing: {missing}, unique categories: {unique_count}, most common: {most_common}")
````

## Activity 3 — One Function, a Default Parameter, and a Decision

**Task:** Write a function `profile_many(df, *column_names, decimals=2)` that accepts any number of numeric column names and returns a summary for each one, using your `profile_numeric()` function from Activity 1.

This uses the same `*args pattern as add_all(*numbers)` — except now each "argument" is a column name instead of a number.

````python
# Solution:
def profile_many(df, *column_names, decimals=2):
    for column_name in column_names:
        count, mean_val, min_val, max_val = profile_numeric(df, column_name)
        print(f"{column_name} -> count: {count}, mean: {round(mean_val, decimals)}, min: {min_val}, max: {max_val}")

# Test:
profile_many(df, 'BMI', 'Age', 'Fasting_Glucose_mg_dL')
````
